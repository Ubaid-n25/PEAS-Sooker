# Task
Create a Python simulation of a soccer agent that moves to take possession of a ball and kicks it into a goal. The script should define a `SoccerAgent` class, manage states such as 'seeking_ball' and 'in_possession', and include a simulation loop that updates and prints the positions of the agent, ball, and goal at each time step until a goal is scored.

## Initialize Simulation Environment

### Subtask:
Set up the initial world state and constants for the soccer simulation.


**Reasoning**:
I will initialize the environment constants and starting positions for the agent, ball, and goal, and set up state variables for the simulation.



```python
import numpy as np

# 1. Define initial 2D coordinates
# Field dimensions: 100x60 (meters)
agent_pos = np.array([10.0, 30.0])
ball_pos = np.array([50.0, 30.0])
goal_pos = np.array([100.0, 30.0])  # Fixed target location

# 2. Define constants
MOVE_SPEED = 2.5  # Meters per step
POSSESSION_THRESHOLD = 1.5  # Distance to pick up the ball
SCORING_THRESHOLD = 2.0  # Distance to goal to consider a score

# 3. Initialize environment state variables
game_state = {
    'goal_scored': False,
    'has_possession': False,
    'steps_taken': 0
}

print(f'Simulation Initialized.')
print(f'Agent Start: {agent_pos}, Ball Start: {ball_pos}, Goal Target: {goal_pos}')
```

## Define Soccer Agent Class

### Subtask:
Create a Python class named `SoccerAgent` to represent the AI agent, including attributes for state management and methods for movement and interaction.


**Reasoning**:
I will define the `SoccerAgent` class with the required attributes and methods for movement, distance calculation, and state management as specified in the instructions.



```python
class SoccerAgent:
    def __init__(self, start_pos, ball_pos):
        self.position = np.array(start_pos, dtype=float)
        self.ball_pos = np.array(ball_pos, dtype=float)
        self.state = 'seeking_ball'

    def get_distance(self, pos1, pos2):
        return np.linalg.norm(pos1 - pos2)

    def move_towards(self, target_pos):
        direction = target_pos - self.position
        distance = np.linalg.norm(direction)
        if distance > 0:
            # Normalize and step
            unit_direction = direction / distance
            self.position += unit_direction * MOVE_SPEED

    def update_state(self):
        dist_to_ball = self.get_distance(self.position, self.ball_pos)
        if self.state == 'seeking_ball' and dist_to_ball <= POSSESSION_THRESHOLD:
            self.state = 'in_possession'
            print("Agent has acquired possession of the ball!")

    def kick_towards_goal(self, goal_pos):
        if self.state == 'in_possession':
            # When in possession, ball moves with the agent or is kicked
            direction_to_goal = goal_pos - self.ball_pos
            dist_to_goal = np.linalg.norm(direction_to_goal)
            if dist_to_goal > 0:
                unit_dir = direction_to_goal / dist_to_goal
                # Move both agent and ball towards goal
                move_step = unit_dir * MOVE_SPEED
                self.position += move_step
                self.ball_pos += move_step

print('SoccerAgent class defined.')
```

## Implement Agent Logic and Simulation Loop

### Subtask:
Develop the core simulation loop to update the agent's actions and display the simulation progress until a goal is scored.


**Reasoning**:
I will implement the core simulation loop by instantiating the SoccerAgent and running a while loop that updates the agent's actions and prints its progress until the ball reaches the goal.



```python
agent = SoccerAgent(agent_pos, ball_pos)

print('Starting Simulation...')

while np.linalg.norm(agent.ball_pos - goal_pos) > SCORING_THRESHOLD:
    game_state['steps_taken'] += 1

    agent.update_state()

    if agent.state == 'seeking_ball':
        agent.move_towards(agent.ball_pos)
    elif agent.state == 'in_possession':
        agent.kick_towards_goal(goal_pos)

    print(f"Step {game_state['steps_taken']}: Agent at {agent.position.round(2)}, Ball at {agent.ball_pos.round(2)}")

game_state['goal_scored'] = True
print(f"\nGOAL SCORED! Total steps taken: {game_state['steps_taken']}")
```
